## Модификации бинарного поиска

```python
def bin_search(arr: List[int], n: int, x: int) -> int:
    l, r = 0, n - 1
    while l <= r:
        mid = (r + l) // 2
        if arr[mid] == x:
            return mid
        elif arr[mid] < x:
            l = mid + 1
        else:
            r = mid - 1
    return -1
```

```python
bin_search([0, 1, 3, 4, 5, 6], 7, -1)
```

### Правый бинарный поиск

```python
def upper_bound(arr: List[int], n: int, x: int) -> int:
    l, r = 0, n
    while l < r:
        mid = (r + l) // 2
        if arr[mid] <= x:
            l = mid + 1
        else:
            r = mid
    return r
```

```python
arr = [1, 2, 2, 3, 4, 5, 5, 6]
x = 2
upper_bound(arr, len(arr), x)
```

### Левый бинарный поиск

```python
def lower_bound(arr: List[int], n: int, x: int) -> int:
    l, r = 0, n
    while l < r:
        mid = (r + l) // 2
        if x <= arr[mid]:
            r = mid
        else:
            l = mid + 1
    return l
```

```python
arr = [1, 2, 2, 3, 4, 5, 5, 6]
x = 2
lower_bound(arr, len(arr), x)
```

## Бинарный поиск по ответу

```python
def sqrt(x: int) -> int:
    left, right = 0, x
    while left < right:
        mid = (left + right + 1) >> 1
        if mid <= x // mid:
            left = mid
        else:
            right = mid - 1
    return left
```

```python
sqrt(17)
```

Задача: Петя активный малый и участвует во всех олимпиадах по математике и физике. Накопил
кучу дипломов, которые лежали в столе и он не знал что с ними делать. И вот он решил, чтобы они перестали пылиться в столе, лучше чтобы они пылились на стене. Для этого ему хотелось их разместить на квадратной доске. Итак, есть 9 прямоугольных дипломов (3Х4), которые надо разместить на квадратной поверхности. Необходимо найти минимальную сторону квадрата для размещения всех дипломов

```python
def bin_search_ans(w: int, h: int, n: int) -> int:
    l = max(w, h)
    r = n * l
    while l + 1 < r:
        mid = (r + l) // 2
        res = (mid // w) * (mid // h)
        if res < n:
            l = mid
        else:
            r = mid
    return r
```

```python
bin_search_ans(3, 4, 9)
```

### Количество траекторий Кузнечика на прямой
Ходим +1 и +2

```python
def hopper(n: int) -> int:
    a0 = 0
    a1 = 1
    if n == 0:
        return 0
    if n == 1:
        return 1
    else:
        res = 0
        for _ in range(n):
            res = a0 + a1
            a0, a1 = a1, res
        return res
```

```python
hopper(10)
```

```python
def hopper3(n: int) -> int:
    if n == 0:
        return 1
    if n == 1:
        return 1
    if n == 2:
        return 2
    n3 = n2 = 1
    n1 = 2
    res = 0
    for _ in range(3, n+1):
        res = n1 + n2 + n3
        n1, n2, n3 = res, n1, n2
    return res
```

```python
hopper3(9)
```

### Траектория наименьшей стоимости для Кузнечика

```python
def hopper_min(values: List[int]) -> int:
    n = len(values)
    res = 0
    a0 = values[0]
    a1 = values[1]
    for i in range(2, n):
        res = values[i] + min(a0, a1)
        a0, a1 = a1, res
    return res
```

```python
values = [3, 4, 2, 5, 1, 7]
hopper_min(values)
```

### Восстановление траектории Кузнечика

```python
def hopper_min2(values: List[int]) -> List[int]:
    n = len(values)
    res = [0] * n
    res[0] = values[0]
    res[1] = values[1]
    for i in range(2, n):
        res[i] = values[i] + min(res[i-1], res[i-2])
    return res
```

```python
def min_trajectory(lst: List[int]) -> List[int]:
    n = len(lst)
    ans = [n-1]
    i = n - 1
    while i > 0:
        ind = i-1 if lst[i-1] == min(lst[i-1], lst[i-2]) else i-2
        ans.append(ind)
        i -= 1 if ind == i else 2
    return ans[::-1]
```

```python
values = [0, 3, 4, 2, 5, 1]
res = hopper_min2(values)
min_trajectory(res)
```

### Рюкзак 0-1, (Жадно)

```python
def knapsack(arr: List[int], x: int) -> int:
    arr = sorted(arr, key=lambda y: y[0] / y[1], reverse=True)
    value = 0
    # print(arr)
    for i in arr:
        if i[1] <= x:
            x -= i[1]
            value += i[0]
    return value
```

```python
arr = [(7, 1), (3, 3), (1, 2), (2, 6), (2, 5)] # <-- не правильный ответ
x = 10
knapsack(arr, x)
```

### Рюкзак 0-1 (Рекурсия)

```python
def knapsack_recursive(W: int, arr: List[int], n: int):
    if n == 0 or W == 0:
        return 0
    
    if arr[n-1][1] > W:
        return knapsack_recursive(W, arr, n-1)
    else:
        return max(
            arr[n-1][0] + knapsack_recursive(W - arr[n-1][1], arr, n-1),
            knapsack_recursive(W, arr, n-1)
        )
```

```python
arr = [(7, 1), (3, 3), (1, 2), (2, 6), (2, 5)] # выдает правильный ответ
x = 10
knapsack_recursive(x, arr, len(arr))
```

### Рюкзак 0-1 (Рекурсия с мемоизацией)

```python
def knapsack_memo(W: int, arr: List[int], n: int, t: List[int]) -> int:
    if n == 0 or W == 0:
        return 0
    
    if t[n][W] != -1:
        return t[n][W]
    
    if arr[n-1][1] <= W:
        t[n][W] = max(
            arr[n-1][0] + knapsack_memo(
                W-arr[n-1][1], arr, n-1, t),
            knapsack_memo(W, arr, n-1, t)
        )
        return t[n][W]
    elif arr[n-1][1] > W:
        t[n][W] = knapsack_memo(W, arr, n-1, t)
        return t[n][W]
```

```python
arr = [(7, 1), (3, 3), (1, 2), (2, 6), (2, 5)]
x = 10
n = 5
t = [[-1 for i in range(x + 1)] for j in range(n + 1)]
knapsack_memo(x, arr, len(arr), t)
```
